```python
import pandas as pd 

df = pd.read_csv(
    r"C:\Users\thana\Desktop\flight_analysis\data\DelayedFlights.csv"
)

print("Γραμμές:", len(df))
print("Στήλες:", len(df.columns))
print(df.dtypes)
```

```python
null_counts = df.isnull().sum()
print(null_counts)
```

```python
#Delay cols με 0
delay_cols = ["CarrierDelay", "WeatherDelay", "NASDelay",
              "SecurityDelay", "LateAircraftDelay"]
df[delay_cols] = df[delay_cols].fillna(0)

#Διαγραφή γραμμών με 0
df = df.dropna(subset=["ArrDelay", "ArrTime", "AirTime",
                       "ActualElapsedTime", "TaxiIn", "TaxiOut"])

print("Γραμμές μετά το καθαρισμό:", len(df))
```

```python
#Target variable: πτήση "delayed" αν ArrDelay > 30'
df["DELAYED"] = (df["ArrDelay"] > 30).astype(int)

#Ώρα αναχώρησης -> κατηγορία
df["DEP_PERIOD"] = pd.cut(
    df["CRSDepTime"],
    bins=[0, 600, 1200, 1800, 2400],
    labels=["night", "morning", "afternoon", "evening"]
)

#Κατανομή target variable
print(df["DELAYED"].value_counts())
print(f"\nDelayed: {df['DELAYED'].mean()*100:.1f}%")
```

```python
df.to_csv(
    r"C:\Users\thana\Desktop\flight_analysis\data\flight_clean.csv",
    index=False
)

print("Επιτυχής αποθήκευση!")
```
