```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

df = pd.read_csv(r"C:\Users\thana\Desktop\flight_analysis\data\flight_clean.csv")

print("Γραμμές:", len(df))
print("Έτοιμο!")
```

```python
# Cyclical encoding για μήνα
df["Month_sin"] = np.sin(2 * np.pi * df["Month"] / 12)
df["Month_cos"] = np.cos(2 * np.pi * df["Month"] / 12)

print("Month_sin & month_cos προστέθηκαν!")
```

```python
# Αφαίρεση leakage & άχρηστων features
cols_to_drop = ["LateAircraftDelay", "SecurityDelay", 'WeatherDelay']
df = df.drop(columns=cols_to_drop)

print("Αφαιρέθηκαν:", cols_to_drop)
print("Υπολοιπόμενες στήλες:", df.columns.tolist())
```

```python
# Weekend vs Weekday
df["IS_WEEKEND"] = df["DayOfWeek"].isin([6, 7]).astype(int)

print("Κατανομή IS_WEEKEND:")
print(df["IS_WEEKEND"].value_counts())
```

```python
#IQR Capping για ArrDelay
Q1 = df["ArrDelay"].quantile(0.25)
Q3 = df["ArrDelay"].quantile(0.75)
IQR = Q3 - Q1
upper = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}, Upper cap: {upper}")

df["ArrDelay"] = df["ArrDelay"].clip(upper=upper)

print("ArrDelay capped!")
```

```python
df.to_csv(r"C:\Users\thana\Desktop\flight_analysis\data\flight_ml_ready.csv", index=False)

print("Ολοκλήρωση αποθήκευσης!")
print("Shape:", df.shape)
```
